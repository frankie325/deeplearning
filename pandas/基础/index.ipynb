# 创建 Series 对象


```python
import pandas as pd
import numpy as np

# 创建Series对象, 默认自增索引
s1 = pd.Series([1, 2, 3, 4, 5])
print(s1)
```

```python
# 创建自定义索引
s2 = pd.Series([1, 2, 3, 4, 5], index=["a", "b", "c", "d", "e"])
print(s2)
# 字典形式
s3 = pd.Series({"a": 1, "b": 2, "c": 3})
print(s3)
# 使用numpy创建Series对象
s4 = pd.Series(np.arange(5))
print(s4)
```

## Series 对象的属性


```python
# s6 = pd.Series(data=[0,1,2,3,4,5], index=['A', 'B', 'C', 'D', 'E', 'F'])
# print(s6)

# 加入列表推导式
s6 = pd.Series(data=[i for i in range(6)], index=[i for i in "ABCDEF"])
print(s6)

# 获取Series对象的 索引列（的值）
print(s6.index)
# 获取Series对象的 值列（的值）
print(s6.values)
# Series支持根据索引获取元素，即Series[索引值]
print(s6["D"])
# 根据索引，修改Series对象的元素值
s6["D"] = 99
print(s6)
```

# DataFrame 对象入门

## 创建 DataFrame 对象


```python
# 场景1：通过 字典+列表 来创建DataFrame对象
# 每个键值对 = 1列数据
info = {
    "name": ["水冷哥", "深情哥", "紫琪", "德华"],
    "gender": ["男", "男", "女", "男"],
    "age": [25, 30, 22, 28],
}

# 把上述数据集封装成DataFrame对象
df1 = pd.DataFrame(data=info)
print(df1)

# 场景2：通过 列表 + 元组
#  每个元组 = 1行数据

info = [
    ("刘亦菲", "女", 35),
    ("迪丽热巴", "女", 31),
    ("王志奇", "未知", 66),
]
# columns表示列名
df2 = pd.DataFrame(data=info, columns=["name", "gender", "age"])
print(df2)

# 场景3：通过numpy的ndarray -> pandas DataFrame 的方式实现
arr1 = np.arange(12).reshape(3, 4)
df3 = pd.DataFrame(data=arr1, columns=["a", "b", "c", "d"])
print(df3)
```

# pandas基本数据操作

```python
data = pd.read_csv("./data/stock_day.csv")
print(data)

# 移除不需要的列
data.drop(columns=["ma5", "ma10", "ma20"], axis=0, inplace=True)  # 0表示列，1表示行
print(data)
```

## 索引操作

```python
# 先列后行，没有先行后列的取值方式
print(data['open']) # Series
print(data["open"][0])  # 23.53
```

```python
# loc索引：行索引和列名来获取元素
print(data.loc["2018-02-27", "open"])  # 23.53

# print(data.loc["2018-02-27":"2018-02-14", :]) # :表示所有列
print(data.loc["2018-02-27":"2018-02-14", ["open", "high"]])


# iloc索引：行号和列索引来获取元素
print(data.iloc[0, 0])  # 23.53
print(data.iloc[0:4, 0:5]) # 第0行到第3行，第0列到第4列
```

```python
# 赋值操作
# data['open'] = 23
# data.low = 3
# print(data)
```

# 排序操作


```python
# 基于open列进行排序
print(data.sort_values(by="open", ascending=True)) # 升序

# 基于open列进行降序，再基于high列进行降序
print(data.sort_values(by=["open", "high"], ascending=[False, False]))
```

```python
# 按照索引排序
print(data.sort_index(ascending=True))

# Series对象也有排序方法
# print(data.open.sort_index(ascending=True)) # 对open列的索引进行排序
print(data.open.sort_values(ascending=True)) # 对open列的值进行排序
```

# DataFrame的运算

```python
# 对close列的每个元素进行加2操作
data["close"] + 2 
# data.close.add(2) # 效果同上
print(data)

# 针对low列的每个元素进行减10操作
data["low"] - 10
# data.low.sub(10) # 效果同上
print(data)
```

```python
# 逻辑运算符

# 筛选出open列大于23的值
data[data.open > 23]

# 筛选出open列大于23且close列小于25的值
data[(data.open > 23) & (data.close < 25)]
```

```python
# 可以通过query函数优化上面的代码
data.query("open > 23 and close < 25")
```

```python
# 固定值的筛选，isin
# data[(data.open == 23.53) | (data.open == 23.67)]
# data.query("open in [23.53, 23.67]")
data[data.open.isin([23.53, 23.67])]
```

## 统计函数

```python
# 查看DataFrame的统计信息
print(data.describe()) 

# 对每列进行求和
print(data.sum())
# 对open列进行求和
print(data.open.sum()) 
print(data.mean()) # 均值
print(data.median()) # 中位数
print(data.cumsum()) # 累计和
```

## 自定义函数

```python
# 定义一个对列，最大值-最小值的函数
data[['open','close']].apply(lambda x: x.max() - x.min(),axis=0)
```
