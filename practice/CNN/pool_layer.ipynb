```python
"""
案例:
    演示池化层相关操作.

池化层解释(Pooling):
    目的:
        降维.
    思路:
        最大池化.
        平均池化.
    特点:
        池化不会改变数据的 通道数.
"""
```

```python
from this import d
import torch
import torch.nn as nn

def dm01():
    # 1. 创建1个 1通道 3*3的二维矩阵.
    # inputs = torch.tensor([ # 1 通道C
    #     [                   # 3 高度H
    #         [0, 1, 2],      # 3 宽度W
    #         [3, 4, 5],
    #         [6, 7, 8]
    #     ]
    # ],dtype=torch.float32)

    # 多通道池化
    inputs = torch.tensor([     # 3 通道C
        [                       # 通道1, HW 3,3
            [0, 1, 2],
            [3, 4, 5],
            [6, 7, 8]
        ],

        [                       # 通道2, HW 3,3
            [10, 20, 30],
            [40, 50, 60],
            [70, 80, 90]
        ],

        [                       # 通道3, HW 3,3
            [11, 22, 33],
            [44, 55, 66],
            [77, 88, 99]
        ]
    ],dtype=torch.float32)

    # 1. 最大池化
    # 参1: 池化核(池化窗口)大小, 参2: 步长, 参3: 填充.
    pool1 = nn.MaxPool2d(kernel_size=2, stride=1, padding=0)
    outputs = pool1(inputs)
    print(f"outputs: {outputs}, shape: {outputs.shape}") # (1, 2, 2) 多通道：(3, 2, 2)

    # 2. 平均池化
    pooling = nn.AvgPool2d(kernel_size=2, stride=1, padding=0)
    output = pooling(inputs)
    print(f"output: {output}, shape: {output.shape}") # (1, 2, 2) 多通道：(3, 2, 2)

if __name__ == '__main__':
    dm01()
```
