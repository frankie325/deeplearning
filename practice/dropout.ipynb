```python
import torch
import torch.nn as nn


def test():
    # 1.创建隐藏层输出结果
    t1 = torch.randint(0, 10, size=(1, 4)).float()
    print(f"t1:{t1}")
    # 2.创建下一层和激活函数的计算
    linear = nn.Linear(4, 5)
    t2 = linear(t1)
    print(f"t2:{t2}")
    output = torch.relu(t2)
    print(f"output:{output}")  # 激活函数
    # 3.创建dropout层
    dropout = nn.Dropout(p=0.5)  # 每个神经元都有50%的概率被kill
    d1 = dropout(output)
    print(
        f"随机失活后的数据d1:{d1}"
    )  # 未被失活的数据进行缩放, 缩放比例为 1/(1-p)，这里是element * 2


if __name__ == "__main__":
    test()
```
