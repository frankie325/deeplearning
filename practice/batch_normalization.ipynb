```python
"""
BatchNorm1d: 主要应用于全连接层或处理一维数据的网络。它接受形状为（N， num_features）的张量输入，
其中N是批量大小，num_features是每个样本的特征数量。
BatchNorm2d: 主要应用于卷积层或处理二维数据的网络。它接受形状为（N， C， H， W）的张量输入，
其中N是批量大小，C是通道数，H是高度，W是宽度。
BatchNorm3d: 主要应用于处理三维卷积神经网络，处理三维数据或特征图，例如视频数据。它接受形状为（N， C， D， H， W）的张量输入，
其中N是批量大小，C是通道数，D是深度，H是高度，W是宽度。
"""

import torch
import torch.nn as nn


def dm01():
    # 1. 创建图像样本数据
    # 1张图片，2个通道，3行4列（像素点）
    input_2d = torch.randn(size=(1, 2, 3, 4))
    print(f"输入数据input_2d:{input_2d}")

    # 2.创建归一化层（BN层）
    # 参1：输入特征数 = 图片通道数
    # 参2：噪声值（小常数），默认1e-5
    # 参3：动量因子，用于计算移动平均统计量的，默认0.1
    # 参4：表示使用可学习的变换参数（γ， β）对归一化（标准化）后的数据进行缩放和平移，默认True
    bn2d = nn.BatchNorm2d(num_features=2, eps=1e-5, momentum=0.1, affine=True)

    # 3. 对数据进行：批量归一化处理
    output_2d = bn2d(input_2d)
    print(f"输出数据output_2d:{output_2d}")

# 处理一维数据
def dm02():
    # 1. 创建一维样本数据
    # 2个样本，2个特征
    input_1d = torch.randn(size=(2, 2))
    print(f"输入数据input_1d:{input_1d}")

    # 2.创建线性层
    linear = nn.Linear(2, 4)

    # 3.对数据进行线性变换
    l1 = linear(input_1d)

    # 4.创建批量归一化层
    bn1d = nn.BatchNorm1d(num_features=4) # 上层l1输出的是4个特征
    # 5.对线性处理结果l1做批量归一化处理
    output = bn1d(l1)
    print(f"批量归一化处理后的输出数据output:{output}")

if __name__ == "__main__":
    dm01()
    dm02()
```
