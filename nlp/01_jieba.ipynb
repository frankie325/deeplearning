```python
import jieba
import pkg_resources

content = "传智教育是一家上市公司，旗下有黑马程序员品牌。我是在黑马这里学习人工智能"

# 精确模型：试图将句子最精确地切开，适合文本分析。也属于默认模式
seg_list = jieba.cut(content, cut_all=False)  # cut_all默认为False
print(seg_list) # 返回生成器
print([*seg_list])
# print("精确模型：" + "/ ".join(seg_list))
# 若需直接返回列表内容, 使用jieba.lcut即可
print(jieba.lcut(content))

# 全模式分词：把句子中所有的可以成词的词语都扫描出来, 速度非常快，但是不能消除歧义
li = jieba.lcut(content, cut_all=True)  # cut_all默认为False
print(li)

# 搜索引擎模式：在精确模式的基础上，对长词再次切分，提高召回率，适合用于搜索引擎分词。
li = jieba.lcut_for_search(content)
print(li)

# 中文繁体分词：
# content = "煩惱即是菩提，我暫且不提"
li = jieba.lcut(content)
print(li)
```
