# RNN模型

![rnn.png](./assets/images/rnn.png)

```python
import torch
import torch.nn as nn

"""
第一个参数：input_size(输入词向量的维度，--列数)
第二个参数：hidden_size(隐藏层的维度， 隐藏层的神经元个数)
第三个参数：num_layers(隐藏层的数量)
"""
input_size = 5
hidden_size = 6
num_layers = 1
rnn = nn.RNN(input_size, hidden_size, num_layers)

"""
第一个参数：sequence_length(序列的长度，单词的个数)
第二个参数：batch_size(批量大小)
第三个参数：input_size(输入词向量的维度，--列数)

好理解的方式应该是：可以设置RNN(batch_first=True)参数，将批次大小放在第一个维度
（2，3，x）: 每个批次2个句子，每个句子3个分词，每个分词x维向量
[
    句子1：[
        分词1向量：[]
        分词2向量：[]
        分词3向量：[]
    ]
    句子2：[
        分词1向量：[]
        分词2向量：[]
        分词3向量：[]
    ]
]

RNN输入数据默认参数形状为（3，2，x）: 将分词数量放在第一个维度，
[
    分词1：[
        句子1：[]
        句子2：[]
    ]
    分词2：[
        句子1：[]
        句子2：[]
    ]
    分词3：[
        句子1：[]
        句子2：[]
    ]
]
"""

sequence_length = 4
batch_size = 3
input = torch.randn(sequence_length, batch_size, input_size)
print('输入数据-->', input)
```

输入数据的形状含义

![word2vec.png](./assets/images/rnn_params.png)


```python
"""
第一个参数：num_layers（隐藏层的数量）
第二个参数：batch_size(批量大小)
第三个参数：hidden_size(RNN单元输出的隐藏层张量维度)
"""

h0 = torch.randn(num_layers, batch_size, hidden_size)
print("隐藏层初始状态-->", h0)

output, hn = rnn(input, h0)
print("输出数据-->", output)
print("隐藏层最终状态-->", hn)
```
